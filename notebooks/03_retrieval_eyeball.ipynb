# Retrieval — Eyeball Evaluation

Compare Vector, BM25, and Hybrid retrievers on 10 realistic queries.

**Goal:** build intuition for where each retriever wins/loses before formal Ragas eval on Day 6.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import os

from dotenv import load_dotenv
from IPython.display import Markdown, display
from openai import OpenAI

from src.copilot.retrieval.bm25 import BM25Retriever
from src.copilot.retrieval.hybrid import HybridRetriever
from src.copilot.retrieval.vector import VectorRetriever

load_dotenv(Path.cwd().parent / '.env')

db_url = os.environ['DATABASE_URL']
client = OpenAI(api_key=os.environ['OPENAI_API_KEY'])

vec = VectorRetriever(db_url, client)
bm25 = BM25Retriever(db_url)
hybrid = HybridRetriever(db_url, client)
```

```python
QUERIES = [
    'Qual a alíquota do IBS?',
    'Quem é contribuinte da CBS?',
    'Como funciona o split payment?',
    'O que é cashback tributário?',
    'Base de cálculo do IBS em operações com desconto',
    'Quais operações são imunes ao IBS?',
    'Como calcular crédito presumido de CBS?',
    'Regime especial para microempresas',
    'Fato gerador do Imposto Seletivo',
    'Devolução de tributos na exportação',
]
```

```python
def show_results(query: str, k: int = 5):
    md = [f'## {query}\n']
    for name, retriever in [('Vector', vec), ('BM25', bm25), ('Hybrid', hybrid)]:
        md.append(f'### {name}\n')
        results = retriever.retrieve(query, k=k)
        if not results:
            md.append('_no results_\n')
            continue
        for i, r in enumerate(results, 1):
            snippet = r.text[:200].replace('\n', ' ')
            md.append(f'{i}. **{r.article}** · {r.source} · `{r.score:.3f}`  \n   {snippet}...')
        md.append('')
    display(Markdown('\n'.join(md)))
```

```python
for q in QUERIES:
    show_results(q)
```
